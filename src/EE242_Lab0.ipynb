### Part 0.1.1 - Palindromes

#### Part 0.1.1.1 - Palindrome (Code)

Write the python code for reading input from user and write a function to check whether the string is a palindrome using the following algorithm:

1. Read in a string input, called `input_string` (this is given to you)
2. Create a string called `input_string_rev` that reverses the all characters in `input_string`, including any punctuation and spaces (e.g. the reverse of "dog" is "god"). 
3. Write a `for` loop to check that `input_string` and `input_string_rev` are the same string.
4. If `input_string` is the same string as `input_string_rev` return `True`, otherwise return `False`.

````python
a = input("Input A:")

def isPalindrome(input_string: str) -> bool:
    # Your code goes here
    input_string_rev = input_string[::-1]

    for i in range(len(input_string)):
        if input_string[i] != input_string_rev[i]:
            return False
    return True

test_output_b = isPalindrome("Dinuka Sahabandu")
print("Output B (should be False):", test_output_b)  # print() will always print each argument, with spaces between each one.

test_output_c = isPalindrome("Signals, Systems and Data I:I ataD dna smetsyS ,slangiS")
print(f"Output C (should be True): {test_output_c}")  # f-strings substitute in the value of test_output_c into the brackets.

test_output_d = isPalindrome(a)
print("Output D: " + str(test_output_d))  # Using "+" creates a new string that combines the left and right hand sides.
````

#### Part 0.1.1.2 - Palindrome (Report Questions)

**1. Is there a way to make this algorithm more efficient in time required and storage space?**

Space-wise, we can optimize this algorithm by not creating a new string and rather comparing over the existing string. Then speed-wise, rather than comparing over every single character, we only compare until the iterators (forward and reverse) meet in the middle, and that way we cut the comparisons in half and we don't re-check the already passing values.

**2. Are there cases where the above algorithm/code will fail?**

The algorithm will work for every literal case of a palindrome. This means that every literal character including spaces, symbols, numbers, captilizations need to be included when accounting for a palindrome. When we hear about palindromes in standard English, we might make exceptions when it comes to spaces, symbols, punctuation, so in those cases where something like "A man, a plan, a canal: Panama" which we as humans would identify as a palindrome, wouldn't be considered a true palindrome to our code because of the capitilizations, punctuation and spaces.

#### Part 0.2.1.1 (Optional) - Better dependency management using UV.


This is optional but highly recommended. As you advance through your career you may find that installing packages using `pip` is not a wise idea. This is because different projects may require different versions of libraries but pip makes it so every Python project on your PC has to use the same global packages. It would be nice to allow each project to use its own versions of libraries. Python virtual environments typically solve this problem, but they are cumbersome and clunky to use.

UV solves this problem by managing your Python virtual environments (and even Python versions!) for you. UV allows each project folder to have its own dependencies. To get started, make sure your Lab0 is in its own directory and not something like `Documents` or `Desktop`. Once that is done, [install UV](https://docs.astral.sh/uv/#projects).

```

```

This command tells `uv` to setup a project in this directory. Next, install the required packages.

```
uv add numpy scipy matplotlib ipykernel
```

Finally, to use Jupyter with this project, we need to create a new Jupyter associated with this project. 


Exit and reload VS Code. You should see on the top right where it asks to select a kernel the option to use kernel `lab0`.

One disadvantage of using virtual environments is that the autocompletion from Python doesn't know about them by default. We can fix this.
